5. Площадь фигурыЮ ограниченную линиями у = 2х, у = -х2 + 7х - 6

```python
y1 = 2*x
y2 = -x**2 + 7*x - 6

roots = sp.solve(sp.Eq(y1, y2), x)
a, b = sorted(roots, key=lambda r: float(sp.N(r)))

S = sp.integrate(y2 - y1, (x, a, b))
print("S =", sp.simplify(S), "≈", float(S))
```

```python
xs = np.linspace(float(a)-1, float(b)+1, 400)
plt.figure(figsize=(6,6))
plt.plot(xs, 2*xs, label="y=2x")
plt.plot(xs, -xs**2 + 7*xs - 6, label="y=-x^2+7x-6")
xf = np.linspace(float(a), float(b), 300)
plt.fill_between(xf, -xf**2 + 7*xf - 6, 2*xf, alpha=0.3)
plt.axvline(float(a)); plt.axvline(float(b))
plt.legend(); plt.xlabel("x"); plt.ylabel("y")
plt.tight_layout(); plt.show()
```

6. Количество потребляемой предприятием электроэнергии меняется в течение суток в зависимости от времени t со скоростью v(t) = 8+4sin(pi(t+7)/4), где время t измеряется в часах. Найти суммарный расход электроэнергии за сутки.

```python
v = 8 + 4*sp.sin(sp.pi*(t+7)/4)
E_day = sp.integrate(v, (t, 0, 24))
print("Суммарный расход за сутки =", sp.simplify(E_day), "≈", float(E_day))
print("Средняя скорость =", sp.simplify(E_day/24))
```

7. Найти объём продукции, произведённый за 5 лет, если функция Кобба-Дугласа имеет вид: F(t) = (1 + t)e3t

```python
F = (1 + t)*sp.exp(3*t)
Q = sp.integrate(F, (t, 0, 5))
print("Объём за 5 лет =", sp.simplify(Q), "≈", float(Q))
```
