```python
import pandas as pd
import matplotlib.pyplot as plt
#есть вариант выводить данные в формате xslx, тогда не придётся использовать encoding и разделитель sep
#но данные храняться в csv формате. Посмотреть тип разделителя можно через блокнот - открыть файл csv в блокноте

csv_data = pd.read_csv('Результаты олимпиады.csv', sep=';',encoding='cp1251')
print(csv_data.head(10))
#прежде чем заполнять пустые строки нулями, необходимо узнать, какой вообще тип данных содержит каждый столбец

csv_data.info()

#таблица не такая большая, поэтому можно визуально оценить, где есть пропущенные значения
#Non-null count - количество непустых значений в столбце. 77 - entries - количество строк в целом (но +1 т.е 78 строк, потому что считаются индексы).

csv_data['A'] = csv_data['A'].fillna(0)
csv_data['B'] = csv_data['B'].fillna(0)
csv_data['C'] = csv_data['C'].fillna(0)
csv_data['D'] = csv_data['D'].fillna(0)
csv_data['E'] = csv_data['E'].fillna(0)
csv_data['F'] = csv_data['F'].fillna(0)
csv_data['G'] = csv_data['G'].fillna(0)
csv_data['H'] = csv_data['H'].fillna(0)
#можно было применить ко всему csv_data.fillna(0)
#или с разными значениями по столбцам: df.fillna({'A': 0, 'B': 1}).

print(csv_data.head(10))

#print("Суммарный балл для H:"); print(csv_data['H'].describe())

print("\n");A = csv_data['A'].sum();print("Суммарный балл для A:", A);print("\n");B = csv_data['B'].sum();print("Суммарный балл для B:", B)
print("\n");C = csv_data['C'].sum(); print("Суммарный балл для C:", C);print("\n");D = csv_data['D'].sum();print("Суммарный балл для D:", D)
print("\n");E = csv_data['E'].sum();print("Суммарный балл для E:", E);print("\n");F = csv_data['F'].sum();print("Суммарный балл для F:", F)
print("\n");G = csv_data['G'].sum();print("Суммарный балл для G:", G);print("\n");H= csv_data['H'].sum();print("Суммарный балл для H:", H)

```

```python
data = pd.Series(data = [A, B, C, D, E, F, G, H],index=['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H']); data = data.sort_values(ascending=False)

plt.xlabel('Индекс участника')
plt.ylabel('Суммарные баллы')
data.plot(kind='bar', title='Суммарные баллы по каждому столбцу')
```

```python
Itog1 = csv_data['Итого   I тур'].sum()
Itog2 = csv_data['Итого   II тур'].sum()
print("Итог для I тура: ", Itog1); print("Итог для II тура: ", Itog2)
#if(Itog1 > Itog2){print("Тур I был сложнее, чем тур II")}else if(Itog1 < Itog2){print("Тур II был сложнее, чем тур I")}
print("\n")
if Itog1 > Itog2:
    print("Тур II был сложнее, чем тур I")
elif Itog2 > Itog1:
    print("Тур I был сложнее, чем тур II")
else:
    print("Туры были одинаковой сложности")
```

```python
csv_data = csv_data.sort_values(by='Общий итог',ascending=False)
print(csv_data.head(15))
```

```python
csv_data.plot(x='Фамилия, имя',y='Общий итог', kind='bar',title='Общий итог участников',figsize=(18, 6))
```

```python
csv_data.dropna(axis=0, subset=['Unnamed: 15'], inplace=True)

filtered = csv_data[csv_data['Unnamed: 15'].isin(['Победитель', 'Призер'])]

school_counts = filtered['Школа'].value_counts()

school_counts.plot(kind='bar',title='Количество победителей и призеров по школам',figsize=(12, 6))
plt.xlabel('Школа')
plt.ylabel('Количество')
plt.show()
```
