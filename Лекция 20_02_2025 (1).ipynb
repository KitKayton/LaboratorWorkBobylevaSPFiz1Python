## Дифференциальные уравнения

### Обыкновенные дифференциальные уравнения

 Дифференциальным уравнением называют уравнение, которое связывает функцию одной или нескольких переменных, эти переменные и производные (или дифференциалы) данной функции различных порядков по этим переменным.
 
 Обыкновенное дифференциальное уравнение (включающее функцию одной переменной) имеет вид:
 
 **F(x,y,y',...,y^n) = 0.**
 
 Порядок старшей производной (или дифференциала) функции, входящей в уравнение, определяет порядок дифференциального уравнения.

 Для решения дифференциальных уравнений используется функция dsolve () библиотеки sympy

# dsolve(eq, func, hint='default')

 Решается обыкновенное дифференциальное уравнение **eq** для функции **func** методом **hint**. 
 Параметр **hint** - не обязательный.

```python
#from sympy import Function, dsolve, Eq, Derivative, sin, cos, symbols, exp, sqrt
from sympy import *
import math as m 
```

```python
x = symbols('x')

yf = Function("y")(x)
eq = Eq(diff(yf,x) - (exp(sqrt(x)-2)/sqrt(x)))
dsolve(eq, hint = "separable",simplify = False)
```

```python
x = symbols('x ')
z = exp(sqrt(x)-2)/sqrt(x)
integrate(z,x).simplify()
```

```python
from sympy import *
y = symbols('y')
x = Function('x')
eq = diff(x(y),y) - 2*x(y)/y - y**2 
dsolve(eq, x(y))
```

```python
from sympy import *
u = symbols('u')
v = Function('v')
eq = diff(v(u),u) - v(u)/u - 1/2 
dsolve(eq, v(u))
```

```python
from sympy import *
x = symbols('x')
y = Function('y')
eq = x*diff(y(x),x)-y(x) - x**3 
dsolve(eq, y(x))
```

```python
# Решаем однородное уравнение
from sympy import *
x = symbols('x')
y = Function('y')
eq = diff(y(x),x) + 4*x*y(x)  
dsolve(eq, y(x))
```

Общее решение однородного уравнения  

```python
from sympy import exp, sqrt, diff
integrate(6*x*exp(x**2),x)
```
