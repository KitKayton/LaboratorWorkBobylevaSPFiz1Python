```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator

x = np.arange(0, 11, 1)
y1 = 4 * x
y2 = x**2

fig, ax = plt.subplots(figsize=(8, 6))
# fig - общий контейнер графика (вся область рисунка) 
# ax - отдельная система координат внутри fig, в которой строится график
# по аналогии - fig (лист бумаги), а ax - рамка с осями внутри

ax.scatter(x, y1, c='red', label='y1 = 4*x')
ax.plot(x, y2, label='y2 = x^2')
# эти методы относятся к объекту осей ax, потмоу что именно оси знают, где и как рисовать
# данные. ax.scatter() - рисует отдельные точки (диаграмму рассеяния), ax.plot() - соеди
#няет точки линией (по умолчанию - сплошная линия)

ax.set_xlim(-0.5, 10.5)
ax.set_ylim(-5, 105)

ax.xaxis.set_major_locator(MultipleLocator(2))
ax.xaxis.set_minor_locator(MultipleLocator(0.5))
ax.yaxis.set_major_locator(MultipleLocator(20))
ax.yaxis.set_minor_locator(MultipleLocator(5))
# ax.xaxis — объект оси X, содержащий свойства шкалы и делений.
# set_major_locator определяет, где располагать основные деления.
# set_minor_locator — где располагать промежуточные (минорные).
# MultipleLocator(2) ставит деления каждые 2 единицы.

ax.grid(which='major', linewidth=1.2)
ax.grid(which='minor', linestyle='--', color='gray', linewidth=0.5)
# включает сетку
# which='major' — только по основным делениям.
# which='minor' — по дополнительным.

ax.tick_params(which='major', length=10, width=2)
ax.tick_params(which='minor', length=5, width=1)
# Настраивает внешний вид засечек (тиков).
# which выбирает тип делений.
# length — длина засечки, width — её толщина.

ax.set_title('Графики зависимостей: y1=4*x, y2=x^2')
ax.set_xlabel('x')
ax.set_ylabel('y1, y2')

ax.legend()
# Создаёт легенду, собирая все элементы, у которых указан label.
# Без label легенда элемент не включает.
# Легенда размещается в верхнем правом углу по умолчанию, можно задать loc='lower right' и др.

plt.show()
```
