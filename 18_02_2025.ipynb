##  Интегралы

###  Неопределенный интеграл

**integrate (f(x), х)**
 
 f(x) - подынтегральная функция
 
 х - пе ременная интегрирования

```python
from sympy import *
x = symbols('x')
y=integrate(6*x**5, x)
print (y)
```

### Определенный интеграл

**integrate (f(x), (x, a, b))**

где a, b нижняя и верхняя границы интегрирования.

```python
from sympy import * 
x = symbols('x') 
y=integrate(6*x**5, (x,0,4))
print(y)
```

```python
integrate(sqrt(x+4),(x,-1,0))
```

### Несобственный интеграл
 Отметим, что если границы интегрирования числа, и подынтегральная функция на отрезке интегрирования непрерывна, то интеграл называется собственным; в противном случае - несобственным. 
 
Определенный интеграл расходится, если он не существует или равен бесконечности. 

Геометрически это означает, что площадь криволинейной трапеции не ограничена.

```python
integrate(x**(-4), (x, 1, oo))
```

```python
 integrate (log(x), (x, 0, 1))
```

### Кратные интегралы

```python
from sympy import * 
y = symbols('y')  
integrate(y**2*x-2*x*y,(y,x,2))
```

```python
integrate(-x**4/3 + x**3 - 4*x/3,(x,-1,2))
```
