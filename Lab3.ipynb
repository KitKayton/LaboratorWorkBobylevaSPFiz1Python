Дан набор данных средств на срочных депозитах в тыс. рублей (каждое наблюдение - один вклад).

```python
# этот код  просто запустите его
np.random.seed(42)
deposits = list(np.round(np.random.normal(300, 90, 1000)))

# вас интересует только переменная deposits
print(deposits)
```

```python
# зададим количество интервалов, в которые мы помещали респондентов
bins = 30

# построим гистограмму

mean_ = np.mean(deposits)
median_ = np.median(deposits)

plt.figure(figsize=(8,4))
plt.hist(deposits, bins=bins, edgecolor='black')
plt.title('Распределение сумм вкладов, тыс. руб.')
plt.xlabel('Сумма вклада, тыс. руб.')
plt.ylabel('Частота')
plt.grid(axis='y', alpha=0.3)
plt.axvline(mean_, color = 'yellow',  linestyle='-', linewidth=2, label=f'mean={mean_:.1f}')
plt.axvline(median_, color = 'red',linestyle='--',  linewidth=2, label=f'median={median_:.1f}')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8,4))
plt.hist(deposits, bins=bins, edgecolor='black', color='skyblue')
plt.title('Распределение сумм вкладов, тыс. руб.')
plt.xlabel('Сумма вклада, тыс. руб.')
plt.ylabel('Частота')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

**Задание 3.** Рассчитайте среднее арифметическое, медиану и моду.

```python
import scipy.stats as st

x = np.array(deposits)

mean_ = np.mean(x)
median_ = np.median(x)
mode_ = st.mode(x, keepdims=False).mode

print('Среднее арифметическое:', round(mean_, 2))
print('Медиана:', median_)
print('Мода:', mode_)
```

**Задание 4**. Рассчитайте среднее квадратическое отклонение (СКО).

```python
std_ = np.std(x, ddof=1)   # ddof=1 — выборочное СКО
print('Среднеквадратическое отклонение (СКО):', round(std_, 2))
```

**Задание 5.** Наёдите 95-процентный доверительный интервал.

```python
n = len(x)
confidence = 0.95
interval = st.t.interval(confidence, df=n-1, loc=np.mean(x), scale=st.sem(x))
print(f'95%-доверительный интервал: {interval[0]:.2f} ... {interval[1]:.2f}')
```

**Задание 6.** Наёдите квантили.

```python
q1, q2, q3 = np.percentile(x, [25, 50, 75])
print('25%-квантиль (Q1):', q1)
print('50%-квантиль (медиана, Q2):', q2)
print('75%-квантиль (Q3):', q3)
print('Интерквартильный размах (IQR):', q3 - q1)
```

**Проверка гипотезы**

Предположим, что вы прочитали в газете, что средний размер депозита составляет 306 тысяч рублей и с помощью выборочных данных, которые у вас есть, вы хотите это проверить.

**Задание 8**. Сформулируйте нулевую и альтернативную гипотезы.

```python
# H0: средний размер вклада = 306 тыс. руб.
# H1: средний размер вклада ≠ 306 тыс. руб.
# Это двусторонний t-тест.
```

**Задание 9**. Вычислите p-value.

```python
mu0 = 306  # гипотетическое среднее
t_stat, p_value = st.ttest_1samp(x, popmean=mu0)

print('t-статистика:', round(t_stat, 3))
print('p-value:', round(p_value, 5))
```

**Задание 10**. При пороговом значении в 10%, сделайте вывод о возможности либо невозможности отвергнуть нулевую гипотезу.

```python
alpha = 0.10
if p_value < alpha:
    print('p < α → отвергаем H0. Средний размер вклада отличается от 306 тыс.')
else:
    print('p ≥ α → нет оснований отвергнуть H0. Средний вклад статистически равен 306 тыс.')
```
