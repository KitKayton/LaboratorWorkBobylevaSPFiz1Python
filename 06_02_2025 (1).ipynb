```python
import numpy as np
from sympy import *
from sympy.abc import x
import matplotlib.pyplot as plt
```

```python
def tangent(y,x0):
# Строит уравнение касательной к графику функции у(х) в точке с абсциссой х0 (х0;у0) - точка касания 
    #x0 = 2
    y0 = y.subs(x, x0)
    x1 = x0 + 1
# Производная функции у при х=х0 
    
    k = diff(y,x).subs(x,x0)
    y1 = y0 + k
    
    
    return Line((x0,y0), (x1,y1))
```

###  5. Интервалы выпуклости и точки перегиба графика функции.

```python
print("х_2: %.3f х_2: %.3f" % \
 (xp2[0].evalf(4), xp2[1].evalf(4)))
```

```python
# Вычисляем третью производную в точках х1 и х2. 
# Для получения приближенного значения используем метод .evalf() 
diff(y1, x, 3).subs(x,xp1[0]).evalf(5)
```

```python
 diff(y1, x, 3).subs(x,xp1[1]).evalf(4)
```

 Третья производная в обеих точках отлична от нуля, следовательно, вторая производная в этих точках меняет знак,
 значит получены точки перегиба.
 В точке x1 вторая производная меняет знак с плюса на минус (так как ее производная отрицательна), 
 следовательно, при х <х1 функция выпукла вниз, а от точки x1 до —3 - выпукла вверх.

В точке х2 вторая производная меняет знак с минуса на плюс (так как ее производная положительна), следовательно, при 
—3 < х < х2 функция выпукла вверх, при х > х2 - выпукла вниз.

 ### Экстремум функции одной и  многих переменных
 
 Для поиска экстремума функции одной и многих переменных используется функция minimize () модуля scipy.optimize или библиотека 
sympy

```python
from scipy.optimize import minimize
```

```python
def f(x):
    return  np.exp(-x) - np.exp(-2*x)
```

```python
x = np.linspace(0.1,4,50)
plt.plot(x, f(x), 'r')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True, linestyle='-', color='0.4')
plt.show()
```

```python
def f_max(x):
    return -(np.exp(-x) - np.exp(-2*x))
res = minimize(f_max, -2)
print('x_max: %.3f f_max: %.3f' % (res.x, f(res.x)))
```

```python
# Завершился ли поиск успешно?
res.success
```

##  Условный экстремум

 Поиск экстремума с использованием функции minimize () библиотеки scipy.

```python
 f = lambda w: w[0] - w[1] + 2
```

```python
#График функций f и g
from mpl_toolkits.mplot3d import Axes3D 
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
x = np.linspace(-1, 1, 50) 
у = np.linspace(-1, 1, 50)
x, у = np.meshgrid(x, у) 
z1 = f((x,y)) 
ax.plot_surface(x, y, z1, alpha=0.4)
                     
# Цилиндр х^2+у^2=1
x = np.linspace(-1, 1, 100) 
z = np.linspace(0, 3, 100) 
xc, zc = np.meshgrid(x, z) 
yc = np.sqrt(1-xc**2) 
ax.plot_surface(xc, yc, zc, alpha=0.3) 
ax.plot_surface(xc, -yc, zc, alpha=0.3)
ax.set_xlabel("X") 
ax.set_ylabel("Y") 
ax.set_zlabel("Z") 
plt.show()
```

Из графика заключаем: минимум, ориентировочно, точка 
(-0,5; 0,5), максимум - ориентировочно, точка (0,5; -0,5)

```python
# Словарь, задающий формулы и тип ограничений
cons = ({'type': 'eq', 'fun': lambda w: w[0]**2 + w[1]**2 - 1})

# Границы в виде: все переменные могут принимать произвольные значения 
bnds = ((None, None), (None, None))
    
# Ищем точку минимума 
res = minimize(f, (-0.8, 0.8), bounds=bnds, constraints=cons)
```

```python
res.x
```

```python
f(res.x)
```

```python
# Ищем точку максимума (как точку минимума функции -f) 
f_max = lambda w: -(w[0] - w[1] + 2)
cons = ({'type': 'eq', 'fun': lambda w: w[0]**2 + w[1]**2 - 1})
bnds = ((None, None), (None, None))
res = minimize(f_max, (0.5, -0.5), bounds=bnds, constraints=cons)
```

```python
res.x
```

```python
f(res.x)
```
